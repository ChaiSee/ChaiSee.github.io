## 2. 다운로드만 하기

영상 URL만 넣으면 최고 화질로 내려받습니다.

```python
URL = 'https://www.youtube.com/watch?v=BaW_jenozKc'  # 원하는 영상 URL로 바꾸세요

path = download(URL, output_dir='downloads')
print('저장됨:', path)
```

## 3. 구간 자르기 (길이 편집)

이미 받은 영상에서 `시작~끝` 구간만 잘라냅니다. 시간은 `초`, `분:초`, `시:분:초` 형식을 지원합니다.

```python
clip = trim(path, start='00:05', end='00:15')
print('잘린 구간:', clip)
```

## 4. 다운로드 + 자르기 한 번에

가장 자주 쓰는 방식입니다. 내려받은 뒤 지정 구간만 남깁니다.

- `keep_full=True` 로 하면 원본(전체) 파일도 남깁니다.
- `copy=True` 로 하면 재인코딩 없이 매우 빠르게 자릅니다(키프레임 단위, 시작점이 약간 어긋날 수 있음).

```python
result = download_and_trim(
    URL,
    start='00:30',
    end='01:45',
    output_dir='downloads',
    keep_full=False,
)
print('완료:', result)
```
